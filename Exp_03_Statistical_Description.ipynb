```python
# Summary statistics (count, mean, std, min, 25%, 50%, 75%, max)
data.describe()
```

```python
# Dataset info (data types, memory usage, non-null counts)
data.info()
```

```python
# Central Tendency: Mean of all numerical attributes
data.mean()
```

```python
# Central Tendency: Median (50th percentile) of all numerical attributes
data.median()
```

```python
# Central Tendency: Mode (most frequent values)
data.mode().iloc[0]
```

```python
# Measures of Dispersion: Standard Deviation
data.std()
```

```python
# Measures of Dispersion: Variance
data.var()
```

```python
# Quantiles / Percentiles (25%, 50%, 75%)
data.quantile([0.25, 0.5, 0.75])
```

```python
# Measures of Shape: Skewness
data.skew()
```

```python
# Measures of Shape: Kurtosis
data.kurtosis()
```

```python
# Pairwise correlation matrix
data.corr()
```
