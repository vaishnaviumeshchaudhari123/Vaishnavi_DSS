```python
# Data preprocessing & missing value treatment
# check Missing Value by record 
data.isna().any()
```

```python
# Handle Missing Values
for col in data.columns:
    if pd.api.types.is_numeric_dtype(data[col]):
        data[col] = data[col].fillna(data[col].median())
    else:
        data[col] = data[col].fillna(data[col].mode()[0])

# Drop non-feature identifier 'id' and auxiliary columns if present
data = data.drop(['id', 'Class', 'Type'], axis=1, errors='ignore')

# Encode categorical features into numeric labels
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
for col in data.select_dtypes(include=['object', 'string']).columns:
    data[col] = le.fit_transform(data[col].astype(str))

data.head()
```

```python
# Independent and Dependent Variables
x = data.drop("Depression", axis=1)
y = data["Depression"]
```

```python
# Splitting of DataSet into train and Test
# Splitting the data into training and testing data sets
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
print('Training set:', x_train.shape)
print('Testing set:', x_test.shape)

# Feature scaling for gradient and distance based models
scaler = StandardScaler()
x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
```

```python
# Logistic Regression
from sklearn.linear_model import LogisticRegression
log = LogisticRegression(max_iter=1000)
log.fit(x_train_scaled, y_train)
```

```python
y_pred1 = log.predict(x_test_scaled)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test, y_pred1)
```

```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, y_pred1)
labels = np.unique(y_test)
cm_df = pd.DataFrame(cm, index=labels, columns=labels)

# Plot confusion matrix using seaborn
plt.figure(figsize=(6, 4))
sns.heatmap(cm_df, annot=True, fmt='d', cmap='Blues', linewidths=1, linecolor='black')
plt.xlabel("Predicted Label")
plt.ylabel("True Label")
plt.title("Confusion Matrix - Logistic Regression")
plt.show()
```

```python
# KNN Classifier
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier()
knn.fit(x_train_scaled[:5000], y_train[:5000])
```

```python
y_pred2 = knn.predict(x_test_scaled)
accuracy_score(y_test, y_pred2)
```

```python
# Support Vector Classifier / Machine (SVC/SVM)
from sklearn import svm
svm_model = svm.SVC()
svm_model.fit(x_train_scaled[:5000], y_train[:5000])
```

```python
y_pred3 = svm_model.predict(x_test_scaled)
accuracy_score(y_test, y_pred3)
```

```python
# Decision Trees Algorithm
from sklearn.tree import DecisionTreeClassifier
dt = DecisionTreeClassifier(random_state=42)
dt.fit(x_train, y_train)
y_pred4 = dt.predict(x_test)
print("Accuracy without pruning:", accuracy_score(y_test, y_pred4))
```

```python
dt_depth = DecisionTreeClassifier(
    max_depth=5,
    random_state=42
)
dt_depth.fit(x_train, y_train)
y_pred_depth = dt_depth.predict(x_test)
print("Accuracy with max_depth pruning:", accuracy_score(y_test, y_pred_depth))
```

```python
from sklearn import tree
plt.figure(figsize=(18, 8))
tree.plot_tree(
    dt_depth,
    feature_names=x.columns,
    class_names=["No Depression", "Depression"],
    filled=True,
    rounded=True,
    max_depth=3,
    fontsize=9
)
plt.title("Decision Tree Visualization (Max Depth = 3 Displayed)")
plt.show()
```

```python
# Random Forest Classifier
from sklearn.ensemble import RandomForestClassifier
rf = RandomForestClassifier(n_estimators=100, max_depth=8, random_state=42)
rf.fit(x_train, y_train)
y_pred5 = rf.predict(x_test)
accuracy_score(y_test, y_pred5)
```
